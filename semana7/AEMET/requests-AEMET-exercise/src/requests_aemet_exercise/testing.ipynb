```python
import requests
import os
import json
from dotenv import load_dotenv

load_dotenv()

aemet_server = "https://opendata.aemet.es/opendata"

url_inventario = "/api/valores/climatologicos/inventarioestaciones/todasestaciones"

url = aemet_server+url_inventario

params={
    "api_key": os.getenv("AEMET_API_KEY")
}
response = requests.get(url, params=params)
print("Estado HTTP:", response.status_code)
print("URL:", response.url)

data = response.json()
print(json.dumps(data, indent=2, ensure_ascii=False))
```
